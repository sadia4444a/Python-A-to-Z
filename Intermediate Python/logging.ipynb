Logging is like adding notes to your code to keep track of what's happening while your program runs.
### Why Use Logging?
    1. Instead of print() statements, use logging.
    2. Logging allows different levels of messages (e.g., info, warning, error) and better control of where the messages are stored (e.g., console, file).


#### DEBUG: Detailed information, typically of interest only when diagnosing problems

#### INFO: Confirmation that things are working as expected

#### WARNING: An indication that something unexpected happened, or indicative of some problem in the near future (e.g."disk space low'). The software is still working as expected

#### ERROR: Due to a more serious problem, the software has not been able to perform some function
#### CRITICAL: A serious error, indicating that the program itself may be unable to continue running

### Basic Logging Setup

```python
import logging
```

```python
logging.basicConfig(level=logging.INFO)

logging.info("Hello, World!")
logging.warning("Hello, World!")
logging.error("Hello, World!")
logging.critical("Hello, World!")
logging.debug("Hello, World!")
```

### Save Logs to a File

```python
logging.basicConfig(
    filename='app2.log',  # Log file name
    filemode='w',        # 'w' to overwrite the file, 'a' to append
    format='%(asctime)s - %(levelname)s - %(message)s',  # Customize log format
    level=logging.DEBUG  # Log everything from DEBUG level upwards
)

logging.info("This log will be saved in a file!")
```

```python
def divide(a, b):
    logging.info(f"Dividing {a} by {b}.")
    if b == 0:
        logging.error("Division by zero!")
        return None
    return a / b

result = divide(10, 2)
logging.info(f"Result: {result}")

result = divide(10, 0)
```

### Advanced Logging: Multiple Handlers

```python
import logging

# Create a logger
logger = logging.getLogger()  # Root logger
logger.setLevel(logging.DEBUG)  # Set the logging level

# Create a file handler
file_handler = logging.FileHandler('app2.log', mode='w')  # 'w' overwrites the log file
file_handler.setLevel(logging.DEBUG)  # Log all levels to the file

# Create a formatter and attach it to the handler
formatter = logging.Formatter('%(asctime)s - %(levelname)s - %(message)s')
file_handler.setFormatter(formatter)

# Clear existing handlers (important in Jupyter to avoid duplicates)
if logger.hasHandlers():
    logger.handlers.clear()

# Add the file handler to the logger
logger.addHandler(file_handler)

# Add a console handler to see logs in the notebook (optional)
console_handler = logging.StreamHandler()
console_handler.setLevel(logging.INFO)
console_handler.setFormatter(formatter)
logger.addHandler(console_handler)

# Write logs
logger.info("This is an info message.")
logger.warning("This is a warning message.")
logger.error("This is an error message.")
```

### Basic Setup for Multiple Loggers

```python
import logging

# Logger for API
api_logger = logging.getLogger('API')
api_logger.setLevel(logging.INFO)

# Logger for Database
db_logger = logging.getLogger('Database')
db_logger.setLevel(logging.DEBUG)

# Handlers
api_handler = logging.FileHandler('api.log', mode='w')
db_handler = logging.FileHandler('db.log', mode='w')

# Formatters
formatter = logging.Formatter('%(asctime)s - %(name)s - %(levelname)s - %(message)s')
api_handler.setFormatter(formatter)
db_handler.setFormatter(formatter)

# Attach handlers to loggers
api_logger.addHandler(api_handler)
db_logger.addHandler(db_handler)

# Log messages
api_logger.info("API is running smoothly.")
db_logger.debug("Connected to the database.")
db_logger.error("Database connection failed!")
```

```python
import logging
import time

# # Set up logging for ML pipeline
# logging.basicConfig(
#     filename="ml_training.log",
#     level=logging.DEBUG,
#     format="%(asctime)s - %(levelname)s - %(message)s"
# )


ml_logger = logging.getLogger("ML Pipeline")
ml_logger.setLevel(logging.DEBUG)
ml_handler = logging.FileHandler("ml_training.log", mode="w")
formatter = logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")
ml_handler.setFormatter(formatter)
ml_logger.addHandler(ml_handler)

def train_model(epochs):
    for epoch in range(1, epochs + 1):
        try:
            ml_logger.info(f"Starting epoch {epoch}.")
            # Simulate training
            time.sleep(1)  # Pretend training takes time
            if epoch == 3:  # Simulate an error at epoch 3
                raise RuntimeError("GPU memory overflow.")
            ml_logger.info(f"Completed epoch {epoch}.")
        except RuntimeError as e:
            ml_logger.error(f"Error during epoch {epoch}: {e}")

# Start training
train_model(6)
```
